### Задача 1.9
> Сколько существует семизначных чисел, состоя-
щих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а циф-
ры 5 и 6 - по 2 раза?

```python
n = 7
n1 = 3
n2 = 2
n3 = 2
P = factorial(n)/(factorial(n1)*factorial(n2)*factorial(n3))
P
```

### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и
5 женщин. Найти вероятность того, что: а) никакие два лица одного
пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар.

а) никакие два лица одного
пола не сядут рядом

```python
P = 2*(factorial(5))**2/factorial(10)
P
```

```python
k = 0
n = 0
for i in permutations("МММММЖЖЖЖЖ",r = 10):
    n+=1
    if "ММ" not in ''.join(i) and "ЖЖ" not in ''.join(i):
        k+=1
print(k/n)
```

б) супруги сядут рядом, если эти мужчины и
женщины образуют 5 супружеских пар.

```python
P = factorial(5)*2**5/factorial(10)
P
```

### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пас-
сажирам продано 7 билетов. Найти вероятности того, что пассажи-
ры попали: а) в два купе; б) в семь купе; в) в три купе.

а) в два купе

```python
x = 9
y = 4
n = x*y
k = 7
C = factorial(n)/(factorial(k)*factorial(n-k))
C1 = factorial(x)/(factorial(2)*factorial(x-2))
C2 = factorial(y*2)/(factorial(k)*factorial(y*2-k))
P = C1*C2/C
round(P, 8)
```

б) в семь купе

```python
C1 = factorial(x)/(factorial(k)*factorial(x-k))
P = C1*4**7/C
round(P, 8)
```

 в) в три купе

```python
C1 = factorial(x)/(factorial(3)*factorial(x-3))
C2 = 6*factorial(4)/(factorial(4)*factorial(4-4))*factorial(4)/(factorial(2)*factorial(4-2))*factorial(4)/(factorial(1)*factorial(4-1))+3*factorial(4)/(factorial(3)*factorial(4-3))*factorial(4)/(factorial(3)*factorial(4-3))*factorial(4)/(factorial(1)*factorial(4-1))+3*factorial(4)/(factorial(3)*factorial(4-3))*factorial(4)/(factorial(2)*factorial(4-2))*factorial(4)/(factorial(2)*factorial(4-2))
P = C1*C2/C
round(P, 8)
```
