### Задача 1.48
>В старинной игре в кости необходимо было для выигрыша
получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения l l очков; б) выигрыша.

```python
n = 0
k11 = 0
k = 0
for a in range(1, 7):
    for b in range(1, 7):
        for c in range(1, 7):
            n += 1
            s = a + b + c
            if s == 11:
                k11 += 1
            if s > 10:
                k += 1
k11/n
```
