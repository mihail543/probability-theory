### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

а) в два купе
* Рассмотрим событие $A$ – пассажиры попали в два купе
* P – вероятность события $A$

```python
x = 9
y = 4
n = x*y
k = 7
P = sc.comb(y*2, k)*sc.comb(x, 2)/(sc.comb(n, k))
print(round(P, 8))
```

б) в семь купе
* Рассмотрим событие $B$ – пассажиры попали в семь купе
* P – вероятность события $B$

```python
C1 = factorial(x)/(factorial(k)*factorial(x-k))
P = sc.comb(x, k)
print(round(P, 8))
```

 в) в три купе
 * Рассмотрим событие $C$ – пассажиры попали в три купе
* P – вероятность события $C$

```python
C1 = sc.comb(x, 3)
C = sc.comb(n,k)
C2 = (6 * sc.comb(4, 4) * sc.comb(4, 2) * sc.comb(4, 1) +
      3 * sc.comb(4, 3) * sc.comb(4, 3) * sc.comb(4, 1) +
      3 * sc.comb(4, 3) * sc.comb(4, 2) * sc.comb(4, 2))
P = C1 * C2 / C
print(round(P, 8))
```

### Задача 1.48
>В старинной игре в кости необходимо было для выигрыша
получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения ll очков; б) выигрыша.
* Рассмотрим событие $A$ – выпало 11 очков
* Рассмотрим событие $B$ – выигрыш (>10 очков)
* P1 – вероятность события $A$
* P2 – вероятность события $B$

```python
n = 0
k11 = 0
k = 0
for a in range(1, 7):
    for b in range(1, 7):
        for c in range(1, 7):
            n += 1
            s = a + b + c
            if s == 11:
                k11 += 1
            if s > 10:
                k += 1
P1 = k11/n
print(P1)
```

```python
P2 = k/n
print(P2)
```
