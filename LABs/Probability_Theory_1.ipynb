### Задача 2.9
> В среднем 20% пакетов акций на аукционах про- даются по первоначально заявленной цене. Найти вероятность того, что из 9 пакетов акций в результате торгов по первоначально заявленной цене: 
1) не будут проданы 5 пакетов;
2) будет продано:
     * а) менее 2 пакетов;
     * б) не более 2 пакетов;
     * в) хотя бы 2 пакета;
     * г) наивероятнейшее число пакетов.

```python
from scipy.stats import binom, poisson, norm

n = 9
p = 0.2
q = 1 - p

# 1)
P1 = binom.pmf(4, n, p)

# 2)
P2a = binom.cdf(1, n, p)          # менее 2, т.е. 0 или 1
P2b = binom.cdf(2, n, p)          # не более 2, т.е. 0,1,2
P2c = 1 - binom.cdf(1, n, p)      # хотя бы 2, т.е. >=2


k = np.arange(n + 1)
pmf = binom.pmf(k, n, p)
mode = k[pmf == pmf.max()]

print(round(P1,4), round(P2a,4), round(P2b,4), round(P2c,4))
print(mode, pmf[mode])
```

* а) 480 предприятий;
* б) наивероятнейшее число предприятий;
* в) не менее 480;
* г) от 480 до 520.

```python
n = 1000
p = 0.5
q = 1 - p
p * n * q # можно использовать локальную ф. Муавра-Лапласа
```

```python
# а)
m = 480
x = (m - n * p) / sqrt(n * p * q)
prob = norm.pdf(x) / sqrt(n * p * q)
print(round(prob, 4))
```

```python
# б)
m = max((n * p + p) // 1, (n * p - q) // 1)
x = (m - n * q) / sqrt(n * p * q)
prob = norm.pdf(x) / sqrt(n * p * q)
print(round(prob, 4))
```

```python
# в)
a = 480
b = 1000
x1 = (a - n * p) / sqrt(n * p * q)
x2 = (b - n * p) / sqrt(n * p * q)
prob = norm.cdf(x2) - norm.cdf(x1)
print(round(prob, 4))
```

```python
# г)
a = 480
b = 520
x1 = (a - n * p) / sqrt(n * p * q)
x2 = (b - n * p) / sqrt(n * p * q)
prob = norm.cdf(x2) - norm.cdf(x1)
print(round(prob, 4))
```
